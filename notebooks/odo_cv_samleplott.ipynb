```python
import os
from openpyxl import Workbook, load_workbook
from openpyxl.chart import ScatterChart, Series, Reference

folder = r"C:\Users\jdr\Desktop\Ødoforsøk_Tåjeodden"
output_file = "odo_cv_samleplot_openpyxl.xlsx"

wb_out = Workbook()
ws_out = wb_out.active
ws_out.title = "Samledata"

chart = ScatterChart()
chart.title = "Ødometer samleplot"
chart.x_axis.title = "I"
chart.y_axis.title = "L"

start_col = 1

for fname in os.listdir(folder):
    if not fname.lower().endswith((".xlsx", ".xlsm", ".xls")):
        continue
    fpath = os.path.join(folder, fname)
    print(f"Processing: {fname}")
    try:
        wb = load_workbook(fpath, data_only=True)
        if "Beregninger" not in wb.sheetnames:
            print(f"Sheet 'Beregninger' not found in {fname}")
            continue
        sht = wb["Beregninger"]
        # Read I16:I5011 and L16:L5011
        x_data = [sht[f"I{row}"].value for row in range(16, 5012)]
        y_data = [sht[f"L{row}"].value for row in range(16, 5012)]
        # Remove None values (keep pairs)
        data = [(x, y) for x, y in zip(x_data, y_data) if x is not None and y is not None]
        if not data:
            print(f"No data in {fname}")
            continue
        # Write to output sheet
        ws_out.cell(row=1, column=start_col, value=f"{fname}_I")
        ws_out.cell(row=1, column=start_col+1, value=f"{fname}_L")
        for i, (x, y) in enumerate(data, start=2):
            ws_out.cell(row=i, column=start_col, value=x)
            ws_out.cell(row=i, column=start_col+1, value=y)
        # Add series to chart
        xvalues = Reference(ws_out, min_col=start_col, min_row=2, max_row=i)
        yvalues = Reference(ws_out, min_col=start_col+1, min_row=2, max_row=i)
        series = Series(yvalues, xvalues, title=fname)
        chart.series.append(series)
        start_col += 2
    except Exception as e:
        print(f"Failed {fname}: {e}")

ws_out.add_chart(chart, "E2")
wb_out.save(output_file)
print(f"Samleplot lagret som '{output_file}'")
```

```python
import os
import pandas as pd
from openpyxl import Workbook, load_workbook
from openpyxl.chart import ScatterChart, Series, Reference

folder = r"C:\Users\jdr\Desktop\Ødoforsøk_Tåjeodden"
output_file = "odo_k_samleplot.xlsx"

wb_out = Workbook()
ws_out = wb_out.active
ws_out.title = "Samledata"

chart = ScatterChart()
chart.title = "Ødometer samleplot"
chart.x_axis.title = "Effektiv, gjennomsnittlig aksialspenning, Sigma_av' [kPa]"
chart.y_axis.title = "Permeabilitetskoeffisient, k [m/år]"

start_col = 1

for fname in os.listdir(folder):
    if not fname.lower().endswith((".xlsx", ".xlsm", ".xls")):
        continue
    fpath = os.path.join(folder, fname)
    print(f"Processing: {fname}")
    try:
        wb = load_workbook(fpath, data_only=True)
        if "Beregninger" not in wb.sheetnames:
            print(f"Sheet 'Beregninger' not found in {fname}")
            continue
        sht = wb["Beregninger"]
        # Read I16:I5011 and M16:M5011
        x_data = [sht[f"I{row}"].value for row in range(16, 5012)]
        y_data = [sht[f"M{row}"].value for row in range(16, 5012)]
        # Remove None values (keep pairs)
        data = [(x, y) for x, y in zip(x_data, y_data) if x is not None and y is not None]
        if not data:
            print(f"No data in {fname}")
            continue
        # Write to output sheet
        series_name = fname.split('-02')[1] if '-02' in fname else fname
        ws_out.cell(row=1, column=start_col, value=f"{series_name}_I")
        ws_out.cell(row=1, column=start_col+1, value=f"{series_name}_M")
        for i, (x, y) in enumerate(data, start=2):
            ws_out.cell(row=i, column=start_col, value=x)
            ws_out.cell(row=i, column=start_col+1, value=y)
        # Rolling average (window=10)
        df = pd.DataFrame(data, columns=["I", "M"])
        df["M"] = pd.to_numeric(df["M"], errors="coerce")
        df["M_roll"] = df["M"].rolling(window=10, min_periods=1).mean()
        ws_out.cell(row=1, column=start_col+2, value=f"{series_name}_M_roll")
        for j, y_roll in enumerate(df["M_roll"], start=2):
            ws_out.cell(row=j, column=start_col+2, value=y_roll)
        # Add original series to chart (thin line)
        xvalues = Reference(ws_out, min_col=start_col, min_row=2, max_row=i+1)
        yvalues = Reference(ws_out, min_col=start_col+1, min_row=2, max_row=i+1)
        series = Series(yvalues, xvalues, title=series_name)
        series.graphicalProperties.line.width = 75000  # 0.75 pt
        chart.series.append(series)
        # Add rolling average series to chart (thick line)
        yvalues_roll = Reference(ws_out, min_col=start_col+2, min_row=2, max_row=i+1)
        series_roll = Series(yvalues_roll, xvalues, title=f"{series_name} (rolling avg)")
        series_roll.graphicalProperties.line.width = 150000  # 1.5 pt
        chart.series.append(series_roll)
        start_col += 3
    except Exception as e:
        print(f"Failed {fname}: {e}")

ws_out.add_chart(chart, "E2")
wb_out.save(output_file)
print(f"Samleplot lagret som '{output_file}'")
```
