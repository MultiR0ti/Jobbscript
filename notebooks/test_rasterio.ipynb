```python
import plotly.graph_objects as go
from plotly import data

df = data.gapminder()

df_germany = df.loc[(df.country.isin(["Germany"]))]
df_france = df.loc[(df.country.isin(["France"]))]
df_uk = df.loc[(df.country.isin(["United Kingdom"]))]


df_averages_europe = (
    df.loc[(df.continent.isin(["Europe"]))].groupby(by="year").mean(numeric_only=True)
)
df_averages_americas = (
    df.loc[(df.continent.isin(["Americas"]))].groupby(by="year").mean(numeric_only=True)
)


fig = go.Figure(
    data=[
        go.Scatter(x=df_germany.year, y=df_germany.gdpPercap, name="Germany"),
        go.Scatter(x=df_france.year, y=df_france.gdpPercap, name="France"),
        go.Scatter(x=df_uk.year, y=df_uk.gdpPercap, name="UK"),
        go.Scatter(
            x=df_averages_europe.index,
            y=df_averages_europe.gdpPercap,
            name="Europe",
            legend="legend2",
        ),
        go.Scatter(
            x=df_averages_americas.index,
            y=df_averages_americas.gdpPercap,
            name="Americas",
            legend="legend2",
        ),
    ],
    layout=dict(
        title=dict(
            text="GDP Per Capita"
        ),
        legend=dict(
            title=dict(
                text="By country"
            ),
            xref="container",
            yref="container",
            y=0.65,
            bgcolor="Orange"
        ),
        legend2=dict(
            title=dict(
                text="By continent"
            ),
            xref="container",
            yref="container",
            y=0.85,
            bgcolor="Gold",
            itemwidth=30
        ),
        annotations=[
        dict(
            x=1.05,
            y=0.5,
            xref='paper',
            yref='paper',
            text='<b>Custom Legend</b>',
            showarrow=False,
            font=dict(family="Calibri", size=12, color="black"),
            align='left'
        ),
        dict(
            x=1.05,
            y=0.45,
            xref='paper',
            yref='paper',
            text='Short Name<br>A Much Longer Legend Item Name',
            showarrow=False,
            font=dict(family="Calibri", size=12, color="black"),
            align='left'
        )]
    ),
)

fig.show()
```

```python
import rasterio
from rasterio.plot import show
from rasterio.transform import from_origin
from shapely.geometry import LineString
import random

# Base coordinate
base_coord = (79489.2, 1142126.5)

# Generate a list of 1,000 coordinates around the base coordinate
coords = [(base_coord[0] + random.uniform(-10, 10), base_coord[1] + random.uniform(-10, 10)) for _ in range(1000)]

# Open the raster file
raster_path = r'C:\Users\jdr\Downloads\f-veg_1200_samlet.tif'

# List to store coordinates and their corresponding values
coord_values = []

with rasterio.open(raster_path) as src:
    # Print metadata
    print("Metadata:", src.meta)
    
    # Get the nodata value
    nodata = src.nodata
    print("NoData value:", nodata)

    # Get the raster value at each coordinate
    for coord in coords:
        val = list(src.sample([coord]))[0]
        if val == nodata:
            coord_values.append((coord, None))  # Append None if no data
        else:
            coord_values.append((coord, val[0]))  # Append the value

# Print the first 10 coordinate-value pairs to verify
for coord_val in coord_values[:999]:
    print(coord_val)
```

```python
import ezdxf
dxf_file = r'\\nsv2-nasuni-02\fredrikstad\Prosjekt\O10245\10245026-01\10245026-01-03_ARBEIDSOMRAADE\21_fagomraade\11_Geoteknikk\10245026-09-OMRÅDESTABILITET\Profiler\01_Arcgis3\m_utvekslingmodell_f-vei_f-lark.dxf'

# Load the DWG file
try:
    doc = ezdxf.readfile(dxf_file)
except IOError:
    print("File not found or invalid.")
    exit()
except ezdxf.DXFStructureError:
    print("Invalid or corrupted DXF file.")
    exit()

# Access the Modelspace
msp = doc.modelspace()

# Dictionary to count entity types
entity_counts = {}

# Iterate through all entities in the Modelspace
for entity in msp:
    entity_type = entity.dxftype()
    if entity_type in entity_counts:
        entity_counts[entity_type] += 1
    else:
        entity_counts[entity_type] = 1
        print(f"Entity Type: {entity.dxftype()}")
        # Access all DXF attributes of the entity
        for attr in entity.dxf.all_existing_dxf_attribs():
            print(f"  {attr}: {entity.dxf.get(attr)}")
# Display the entity types and their counts
if entity_counts:
    print("Entities found in the DWG file:")
    for entity_type, count in entity_counts.items():
        print(f"{entity_type}: {count}")
else:
    print("No entities found in the DWG file.")
```

```python
import plotly.graph_objects as go
import numpy as np

# Generate sample data
num_traces = 10
num_points = 100000

fig = go.Figure()

for i in range(num_traces):
    x = np.linspace(0, 10, num_points)
    y = np.sin(x) + np.random.normal(0, 0.1, num_points)
    
    fig.add_trace(go.Scattergl(
        x=x,
        y=y,
        mode='lines',
        name=f'Trace {i+1}'
    ))

# Update layout with legend options
fig.update_layout(
    legend=dict(
        groupclick='toggleitem'  # This controls the behavior when clicking on a group
    ),
    title='Large Dataset with Scattergl',
    xaxis_title='X Axis',
    yaxis_title='Y Axis'
)

fig.show(config={'scrollZoom': True})
```
