```python
import win32com.client
from pathlib import Path


o = win32com.client.Dispatch("Excel.Application")

#o.Visible = False

wb_path = r'\\nsv2-nasuni-02\fredrikstad\Prosjekt\O10245\10245026-01\10245026-01-03_ARBEIDSOMRAADE\21_fagomraade\11_Geoteknikk\10245026-07-FELT-OG LABREGISTRERINGER\Poretrykksmålinger\Opptegninger\Hydrauliske'
pz_updates_path = r'\\\\nsv2-nasuni-02\fredrikstad\Prosjekt\O10245\10245026-01\10245026-01-03_ARBEIDSOMRAADE\21_fagomraade\11_Geoteknikk\10245026-07-FELT-OG LABREGISTRERINGER\Poretrykksmålinger\Opptegninger\Hydrauliske\Grunnvannstand_regneark.xlsx'

updates_wb = o.Workbooks.Open(pz_updates_path)
updates_ws = updates_wb.Worksheets[1]
update_date = updates_ws.Range('A3').Value

# start_cell:
borepoint_cell = 'C3'

while updates_ws.Range(borepoint_cell).Value:
    borepoint_name = updates_ws.Range(borepoint_cell).Value
    dybde_fra_topp_slange = updates_ws.Range(borepoint_cell).Offset(1, 8).Value
    update_date = updates_ws.Range(borepoint_cell).Offset(1, -1).Value
    print(borepoint_name, dybde_fra_topp_slange, update_date)
    borepoint_cell_nr = int(borepoint_cell.replace('C', '')) + 1

    excel_path = Path(wb_path)
    excel_wbs = [str(f) for f in excel_path.glob('*.xlsm') if borepoint_name in str(f) and not '~$' in str(f)][0]
    
    wb = o.Workbooks.Open(excel_wbs)

    ws_index = 2

    ws = wb.Worksheets[ws_index]

    date_cell_start = 'H23'
    date_cell_end = 'H35'
    date_cell_range = [int(date_cell_start.replace('H', '')), int(date_cell_end.replace('H', ''))]

    for date_inx in range(int(date_cell_start.replace('H', '')), int(date_cell_end.replace('H', ''))):
        date = ws.Range('H'+str(date_inx)).Value
        if date == update_date:
            break
        else:
            if not date:
                print(update_date, dybde_fra_topp_slange)
                ws.Range('H'+str(date_inx)).Value = update_date
                ws.Range('AF'+str(date_inx)).Value = dybde_fra_topp_slange
                ws.Range('AF'+str(date_inx)).NumberFormat = "0,00"
                break
        
    
    borepoint_cell = 'C' + str(borepoint_cell_nr)

    wb.Save()
    wb.Close(True)

o.Quit()



    

        


# # pdf-plot:

# pdf_path = r'\\nsv2-nasuni-02\fredrikstad\Prosjekt\O10245\10245026-01\10245026-01-03_ARBEIDSOMRAADE\21_fagomraade\11_Geoteknikk\10245026-07-FELT-OG LABREGISTRERINGER\Poretrykksmålinger\Opptegninger\Hydrauliske\SB1017_testpy.pdf'

# print_area = 'A1:HK70'

# ws.PageSetup.Zoom = False


# ws.PageSetup.PrintArea = print_area

# wb.Worksheets[ws_index].Select()

# wb.ActiveSheet.ExportAsFixedFormat(0, pdf_path)

# wb.Close(True)
# o.Quit()
```
