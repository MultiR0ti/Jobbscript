```python
import os

def convert_tot_to_dtr(input_file, output_file=None):
    """
    Convert TOT file to DTR format.
    Keeps D, A, R columns from original and adds W, AK, AKZ, B, V, AB columns.
    """
    # Read the input file
    with open(input_file, 'r', encoding='utf-8') as f:
        lines = f.readlines()
    
    # Find the separator line (#)
    header_lines = []
    data_lines = []
    found_separator = False
    
    for line in lines:
        line = line.strip()
        if line == '#':
            found_separator = True
            continue
        if not found_separator:
            header_lines.append(line)
        else:
            if line:  # Skip empty lines
                data_lines.append(line)
    
    # Process data lines
    converted_lines = []
    for line in data_lines:
        # Parse the line
        parts = line.split(',')
        data_dict = {}
        for part in parts:
            if '=' in part:
                key, value = part.split('=', 1)
                data_dict[key] = value
        
        # Extract D, A, R values
        D = data_dict.get('D', '0.000')
        A = data_dict.get('A', '0.000')
        R = data_dict.get('R', '0')
        
        # Create new format line
        # W=A, other columns set to 0
        new_line = f"D={D},A={A},W={A},AK=0,AKZ=0,B=0,V=0,AB=0,R={R}"
        converted_lines.append(new_line)
    
    # Determine output file name
    if output_file is None:
        output_file = input_file.rsplit('.', 1)[0] + '.DTR'
    
    # Write the output file
    with open(output_file, 'w', encoding='utf-8') as f:
        # Write header
        for line in header_lines:
            f.write(line + '\n')
        # Write separator
        f.write('#\n')
        # Write converted data
        for line in converted_lines:
            f.write(line + '\n')
    
    print(f"Conversion complete!")
    print(f"Input file: {input_file}")
    print(f"Output file: {output_file}")
    print(f"Converted {len(converted_lines)} data lines")
    
    return output_file

# Convert the file
input_file = file_with_incorrect_format
output_file = convert_tot_to_dtr(input_file)

# Display first few lines of the output
print("\nFirst few lines of output:")
with open(output_file, 'r', encoding='utf-8') as f:
    lines = f.readlines()
    for i, line in enumerate(lines[:10]):
        print(line.rstrip())
```
